```python
import numpy as np
from random import gauss
import matplotlib.pyplot as plt

def f(t,x,u):
    return u*x
def g(t,x,o):
    return o*x
def tray(x0,u,o,n,delta):
    y=np.zeros(n)
    x=x0
    oo=np.sqrt(delta)
    for i in range(n):
        t=0
        while t<1:
            w=oo*gauss(0,1)
            x=x+f(t,x,u)*g(t,x,o)*w
            t=t+delta
        y[i]=x
    return y
u=0.001
o=0.02
x0=10
delta=0.001
n=30
y=tray(x0,u,o,n,delta)
plt.plot(y)
plt.show()
```
